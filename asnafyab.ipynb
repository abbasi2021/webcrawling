```python
from bs4 import BeautifulSoup
import pandas as pd
import requests
import re
import time
```

```python
start_time = time.time()
df=pd.DataFrame(columns=["کلمه کلیدی جستجو","کلمه کلیدی url","نام شغل","مدیریت","توضیحات","شماره تماس","شماره همراه ","منطقه شهرداری ","آدرس ","وب سایت رسمی ","فکس ","ایمیل "])
keyword=input("enter keyword")

r=requests.get("https://asnafyab.ir/search?sw="+keyword)
soup = BeautifulSoup(r.content, 'html.parser')

ul=soup.find("ul", attrs={"class":"pagination"})

if ul!=None:
    print(1111)
    end_page=ul.find_all("a")[-1]
    
    end_page_num=re.findall(r'page=(\d+)',str(end_page))
    end_page_num=str(end_page_num[0])

    for i in range(1,int(end_page_num)+1):
    
        r=requests.get("https://asnafyab.ir/search?sw="+keyword+"&page="+str(i))
        soup = BeautifulSoup(r.content, 'html.parser')    
        h2=soup.find_all("h2",attrs={"class":"JobTitle"})
        print("page :",i)
        
        for hr in h2:
            link=hr.a["href"]

            r=requests.get("https://asnafyab.ir"+str(link))
            soup = BeautifulSoup(r.content, 'html.parser')
            desc=soup.find("div", attrs={"class":"job-desc"})
            a=desc.get_text()

            description=''.join(str(l) for l in a.splitlines()[7:] )   
            job_name=desc.find("div", attrs={"class":"job-name"}).get_text()

            spans=[span for span in desc.find_all("span")]
            mg=spans[0].get_text()
            mg_data=spans[1].get_text()  

            rows=soup.find_all("div", attrs={"class":"job-desc job-contact"})

            l1=rows[0].find_all("div",attrs={"class":"col-sm-3"})
            l2=rows[0].find_all("div",attrs={"class":"col-sm-9"})

            l1_data=[i.get_text() for i in l1]
            l2_data=[i.get_text() for i in l2]
            l1_data=[i.replace(":","") for i in l1_data]
            data=dict(zip(l1_data,l2_data))
            data["توضیحات"]=description
            data["نام شغل"]=job_name
            data["مدیریت"]=mg_data
            link=link.replace("/","")
            data["کلمه کلیدی url"]=link
            data["کلمه کلیدی جستجو"]=keyword
            df = df.append(data, ignore_index=True, sort=False)
            
else:
    print(2222)
    h2=soup.find_all("h2",attrs={"class":"JobTitle"})
    for hr in h2:
        link=hr.a["href"]
        
        r=requests.get("https://asnafyab.ir"+str(link))
        soup = BeautifulSoup(r.content, 'html.parser')
        desc=soup.find("div", attrs={"class":"job-desc"})
        a=desc.get_text()

        description=''.join(str(l) for l in a.splitlines()[7:] )   
        job_name=desc.find("div", attrs={"class":"job-name"}).get_text()

        spans=[span for span in desc.find_all("span")]
        mg=spans[0].get_text()
        mg_data=spans[1].get_text()  

        rows=soup.find_all("div", attrs={"class":"job-desc job-contact"})

        l1=rows[0].find_all("div",attrs={"class":"col-sm-3"})
        l2=rows[0].find_all("div",attrs={"class":"col-sm-9"})

        l1_data=[i.get_text() for i in l1]
        l2_data=[i.get_text() for i in l2]
        l1_data=[i.replace(":","") for i in l1_data]
        data=dict(zip(l1_data,l2_data))
        data["توضیحات"]=description
        data["نام شغل"]=job_name
        data["مدیریت"]=mg_data
        link=link.replace("/","")
        data["کلمه کلیدی url"]=link
        data["کلمه کلیدی جستجو"]=keyword
    

        df = df.append(data, ignore_index=True, sort=False)
print("--- %s seconds ---" % (time.time() - start_time))
```

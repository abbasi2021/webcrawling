```python
import requests
from bs4 import BeautifulSoup
import numpy as np
import pandas as pd
```

```python
team1=[]
team2=[]
results=[]
goal1=[]
goal2=[]
```

```python
n_of_soccer=10
```

```python
#تغییر کد403 به 200
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/60.0.3112.113 Safari/537.36'}
for i in range(1,39):
    url="https://top90.ir/england/premier-league/?tpl=schedules&week={}".format(i)
    r = requests.get('{}#!{}'.format(url, 'meme'), headers=headers)
    
    soup=BeautifulSoup(r.content,'html.parser')
    S2=soup.find_all('dd', class_="mElO1")
    S1=soup.find_all('dd',class_="mElO2")
    result=soup.find_all('dd', class_="mElScore")
    for n in range(0,n_of_soccer):
        team2_name=S2[n].find('a').get_text()
        team2.append(team2_name)
    
        team1_name=S1[n].find('a').get_text()
        team1.append(team1_name)
    
        v=result[n].get_text()
        g1=int(v[0])
        goal1.append(g1)
        g2=int(v[4])
        goal2.append(g2)
    
    
    
    
    
```

```python
soup2=BeautifulSoup(r1.content,'html.parser')
name=soup2.find_all('td', class_="t101")
#بازی ها
p=soup2.find_all('td', class_="t201")
#بردها
w=soup2.find_all('td', class_="t301")
#مساوی
d=soup2.find_all('td', class_="t401")
#باخت
l=soup2.find_all('td', class_="t501")
#گل زده
f=soup2.find_all('td', class_="t601")
#گل خورده
a=soup2.find_all('td', class_="t701")
#تفاضل گل
gd=soup2.find_all('td', class_="t801")
#امتیاز
pts=soup2.find_all('td', class_="t901")
```

```python
#print(pts)
```

```python
team_names=[]
P=[]
W=[]
D=[]
L=[]
F=[]
A=[]
GD=[]
PTS=[]
```

```python
for m in range(0,20):
    names=name[m].find('a').get_text()
    team_names.append(names)
    
    p_=p[m].get_text()
    P.append(p_)
    
    w_=w[m].get_text()
    W.append(w_)
    
    d_=d[m].get_text()
    D.append(d_)
    
    l_=l[m].get_text()
    L.append(l_)
    
    f_=f[m].get_text()
    F.append(f_)
    
    a_=a[m].get_text()
    A.append(a_)
    
    gd_=gd[m].get_text()
    GD.append(gd_)
    
    pts_=pts[m].get_text()
    PTS.append(pts_)
    
```

```python
Ranking=pd.DataFrame({"Team":team_names ,
                     "Rank":PTS ,
                     "GD":GD ,
                     "P":P ,
                     "W":W ,
                     "D":D ,
                     "L":L ,
                     "F":F ,
                     "A":A 
                     })
```

```python
from collections import defaultdict
```

```python
won_last=defaultdict(int)
for index,row in FB.iterrows():
    visitorteam=row["Visitor_Team"]
    hometeam=row["Home_Team"]
    row["VisitorLastWin"]=won_last[visitorteam]
    row["HomeLastWin"]=won_last[hometeam]
    FB.iloc[index]=row
    won_last[hometeam]= row['Homewin']
    won_last[visitorteam]= not row['Homewin']
    
FB.iloc[:15]
```

```python
from sklearn.tree import DecisionTreeClassifier
dtc=DecisionTreeClassifier(random_state=42)

from sklearn.model_selection import cross_val_score
x=FB[["VisitorLastWin","HomeLastWin"]].values
scores=cross_val_score(dtc,x,y,scoring='accuracy',cv=5)
print("accuracy : {0:.2f}".format(100* np.mean(scores)))
```

```python
win_streak=defaultdict(int)

for index,row in FB.iterrows():
    visitorteam=row["Visitor_Team"]
    hometeam=row["Home_Team"]
    row["Visitorwinstreak"]=win_streak[visitorteam]
    row["Homewinstreak"]=win_streak[hometeam]
    FB.iloc[index]=row
    if row["Homewin"]==True:
        win_streak[visitorteam]=0
        win_streak[hometeam]+=1
    else:
        win_streak[hometeam]=0
        win_streak[visitorteam]+=1
        
        
```

```python
FB["HomeTeamGDHigher"]=0

for index,row in FB.iterrows():
    home_team=row["Home_Team"]
    visitor_team=row["Visitor_Team"]
    home_GD=Ranking[Ranking["Team"]==home_team]["GD"].values[0]
    visitor_GD=Ranking[Ranking["Team"]==visitor_team]["GD"].values[0]
    row["HomeTeamGDHigher"]=int(home_GD>visitor_GD)
    FB.iloc[index]=row
```

```python
FB["H_RK"]=0
FB["V_RK"]=0
```

```python
Rk=defaultdict(int)

for index, row in FB.iterrows():
    home_team=row["Home_Team"]
    visitor_team=row["Visitor_Team"]
    if row["Homewin"]==True:
        Rk[home_team]+=3
    
    elif row["H_PTS"]==row["V_PTS"]:
        Rk[home_team]+=1
        Rk[visitor_team]+=1
    else:
        Rk[visitor_team]+=3
    row["H_RK"]=Rk[home_team]
    row["V_RK"]=Rk[visitor_team]
    FB.iloc[index]=row
    
    
        
        
        
```

```python
FB['V_GD']=0
FB['H_GD']=0
```

```python
Goals_scored=defaultdict(int)
Goles_eaten=defaultdict(int)

for index, row in FB.iterrows():
    home_team=row["Home_Team"]
    visitor_team=row["Visitor_Team"]
    Goals_scored[home_team]+=row['H_PTS']
    Goals_scored[visitor_team]+=row['V_PTS']
    Goles_eaten[home_team]+=row['V_PTS']
    Goles_eaten[visitor_team]+=row['H_PTS']
    row['V_GD']=(Goals_scored[visitor_team]-Goles_eaten[visitor_team])
    row['H_GD']=(Goals_scored[home_team]-Goles_eaten[home_team])
    FB.iloc[index]=row
```
